```python
import numpy as np
import scipy.io as sio
import warnings

from dipy.denoise.localpca import mppca

warnings.simplefilter("always", RuntimeWarning)

orig = sio.loadmat("data/meas_gre_dir1.mat")

meas_gre = orig["meas_gre"]
mask_brain = orig["mask_brain"].astype(bool)

print("Original:", meas_gre.shape)
print("Mask:", mask_brain.shape)
```

```python
for level in [10, 30, 50]:

    print(f"\n========== {level}% ==========")

    # noisy data 불러오기
    noise_mat = sio.loadmat(
        f"data/noisy_data/noisy_meas_gre_dir1_{level}.mat"
    )

    noisy_real = noise_mat["noisy_real"].astype(np.float32)
    noisy_imag = noise_mat["noisy_imag"].astype(np.float32)

    # magnitude 생성
    noisy_mag = np.sqrt(
        noisy_real**2 + noisy_imag**2
    )

    print("Input shape:", noisy_mag.shape)
    print("Input NaN:", np.isnan(noisy_mag).sum())
    print("Input Inf:", np.isinf(noisy_mag).sum())

    # MP-PCA
    den = mppca(
        noisy_mag,
        mask=mask_brain,
        patch_radius=2
    )

    # 결과 확인
    print("Output NaN:", np.isnan(den).sum())
    print("Output Inf:", np.isinf(den).sum())

    nan_mask = np.isnan(den).any(axis=-1)

    print("NaN voxel:", nan_mask.sum())
    print(
        "mask 내부 NaN:",
        np.sum(nan_mask & mask_brain)
    )
    print(
        "mask 외부 NaN:",
        np.sum(nan_mask & ~mask_brain)
    )
```
